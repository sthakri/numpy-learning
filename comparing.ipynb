# NumPy vs Python Lists - Performance Comparison

This notebook demonstrates why NumPy is the go-to choice for numerical computing in Python:
- **Speed:** Vectorized operations vs Python loops
- **Memory efficiency:** Compact storage vs object overhead

**Bottom Line:** NumPy can be 10-100x faster than pure Python for numerical operations!

## 1. Speed Comparison: Python Lists

```python
#Using Python lists
import numpy as np
import time
size = 1_000_000

l1 = list(range(size))
l2= list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end= time.time()

print(add[0:10])
print(end-start)
```

## 2. Speed Comparison: NumPy Arrays

```python
#Using Numpy Arrays
import numpy as np
import time
size = 1_000_000

l1 = np.array(list(range(size)))
l2= np.array(list(range(size)))

start = time.time()
add = l1+l2
end= time.time()

print(add[0:10])
print(end-start)
```

## 3. Memory Efficiency Comparison

```python
import sys

list_data = list(range(1000))

numpy_data = np.array(list_data)

print("Python list size:", sys.getsizeof(list_data) * len(list_data), "bytes")
print("NumPy array size:", numpy_data.nbytes, "bytes")
```

## Key Takeaways

- ✅ **NumPy is 10-100x faster** for numerical operations
- ✅ **NumPy uses significantly less memory** (stores only data, not Python objects)
- ✅ **NumPy code is cleaner** and more readable
- ✅ **NumPy is optimized in C** and uses SIMD instructions for parallel processing

**When to use NumPy:** Any numerical computation, especially with large datasets!
