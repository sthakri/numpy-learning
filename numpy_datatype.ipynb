# NumPy Data Types - Memory Optimization

Understanding data types is crucial for:
- **Memory efficiency:** Smaller types = less memory usage
- **Performance:** Proper types enable optimized operations
- **Precision control:** Balance between accuracy and resources
- **Type safety:** Prevent overflow and precision errors

**Key insight:** Choosing the right data type can reduce memory usage by 50-75% without sacrificing functionality!

## 1. Type Conversion and Inference

```python
import numpy as np

arr = np.array([1.5, 2.7, 3.9])
print(arr.dtype)

arr_int = arr.astype(np.int32) # Converting float to int
print(arr_int)

print(arr_int.dtype) 
```

## 2. Memory Efficiency Comparison

```python
#Memory Usage
arr_int64 = np.array([1, 2, 3], dtype=np.int64)
arr_int32 = np.array([1, 2, 3], dtype=np.int32)

print(arr_int64.nbytes) # (3 elements * 8 bytes each)
print(arr_int32.nbytes) # (3 elements * 4 bytes each
```

## Quick Reference Guide

**When to use each type:**
- **int8/uint8:** Small integers, image pixels, ages, percentages
- **int16/uint16:** Medium-range integers, sensor readings
- **int32/int64:** Large integers, IDs, timestamps
- **float16:** When memory is critical (beware of precision loss)
- **float32:** Machine learning, graphics (good balance)
- **float64:** Scientific computing, when precision matters
- **bool_:** Binary flags, masks
- **complex64/128:** Signal processing, quantum computing
